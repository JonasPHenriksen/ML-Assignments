## An analysis/description of which questions are most crucial concerning their placement on the axes.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_excel("alldata.xlsx")
exclude_columns = ['navn', 'parti', 'storkreds', 'alder']
question_columns = [col for col in df.columns if col not in exclude_columns]

X = df[question_columns].copy()
X = X.fillna(X.median()) #Probably better to fill out with something within the party instead of just overall median across the board.

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X) #PCA is sensative to scaling but is not needed since everything is between -2 and 2

pca = PCA(n_components=2)  #Reduce all questions dimensions down to 2 features / components
principal_components = pca.fit_transform(X_scaled)

#PC1: First axis, captures the most variance.
#PC2: Second axis, independent to PC1, captures the next most variance.

loadings = pd.DataFrame(
    pca.components_.T,
    index=question_columns,
    columns=['PC1', 'PC2'] #Each questions is given a weight on PC1 and PC2, how much they influence each
)

#Sort questions by absolute influence on PC1 and PC2
important_PC1 = loadings['PC1'].abs().sort_values(ascending=False)
important_PC2 = loadings['PC2'].abs().sort_values(ascending=False)

print("Top questions influencing Axis 1 (PC1):")
print(important_PC1.head(10))

print("\nTop questions influencing Axis 2 (PC2):")
print(important_PC2.head(10))

#PCA looks at how the answers vary across all candidates and finds the directions (principal components) that capture the most variance. The loadings (or weights) show how much each question contributes to those directions.”
#The way the questions are grouped are by finding a tendency between the answers and the candidates.
```

PC1 (First Axis):

Finds the direction in the data where politicians disagree the most overall.

This is the axis where the answers are most spread out.

It’s a combo of multiple questions — not just one.

main ideological dimension — like economic left vs right, if that's where people differ the most

PC2 (Second Axis):

Finds a second direction, completely perpendicular (orthogonal) to the first.

Captures the next biggest source of disagreement.

This helps separate people who might seem similar along PC1.

Could represent a second dimension, like cultural progressiveness vs conservatism.

| Axis     | Captures…                     | Based On…                              |
| -------- | ----------------------------- | -------------------------------------- |
| **PC1**  | The most variation in answers | A blend of question weights (loadings) |
| **PC2**  | Second-most variation         | Orthogonal (independent) from PC1      |
| Together | Main structure in the data    | Often maps to ideological divides      |


| Term            | What It Is                           | In PCA                          |
| --------------- | ------------------------------------ | ------------------------------- |
| **Eigenvector** | A direction (axis) in the data       | Becomes PC1, PC2, etc.          |
| **Eigenvalue**  | How much variance is along that axis | Becomes % of variance explained |


| Concept                               | Your Understanding                                         | Clarified                                                                                          |
| ------------------------------------- | ---------------------------------------------------------- | -------------------------------------------------------------------------------------------------- |
| **Pattern detection**                 | "PCA finds questions that follow the same pattern"         | ✅ Yes — it finds **co-varying sets of questions**                                                  |
| **Multiple patterns**                 | "There can be multiple patterns"                           | ✅ Yes — each PC captures **one independent pattern** of variation                                  |
| **Strongest first**                   | "It picks the strongest patterns"                          | ✅ Yes — PC1 = strongest, PC2 = second strongest, etc.                                              |
| **Two dimensions = limited patterns** | "We only have two axes"                                    | ✅ Yes — you’re only keeping the **top 2 patterns**, the rest are **discarded**                     |
| **Some questions are left out**       | "Some questions just don’t follow any of the two patterns" | ✅ Correct — they **contribute little or nothing** to PC1 and PC2, and are **minimized** in 2D view |


| Step                     | What It Does                                                    |
| ------------------------ | --------------------------------------------------------------- |
| **Covariance matrix**    | Measures how questions vary together                            |
| **PCA**                  | Picks the **strongest co-varying patterns** (with eigenvectors) |
| **Principal Components** | New axes based on those patterns                                |
| **Eigenvalues**          | Measure how much "impact" (variance) each pattern has           |

## A clustering analysis where you attempt various cluster numbers, which would correspond to different parties. Discuss whether there is room for more clusters/parties or if a reduction is needed. Make sure you cover: **K-Means, Hierarchical clustering, and DBSCAN.**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score
import seaborn as sns

df = pd.read_excel("alldata.xlsx")

exclude_columns = ['navn', 'parti', 'storkreds', 'alder']
question_columns = [col for col in df.columns if col not in exclude_columns]
X = df[question_columns]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

cluster = KMeans(
    random_state=42,
    n_clusters=14,
    n_init=10,
    init='k-means++'
)
y_prime = cluster.fit_predict(X_scaled)
df['Cluster'] = y_prime

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]

unique_parties = df['parti'].unique()
palette = sns.color_palette("hls", len(unique_parties))
party_color_map = {parti: palette[i] for i, parti in enumerate(unique_parties)}

cluster_party_map = (
    df.groupby('Cluster')['parti']
    .agg(lambda x: x.value_counts().index[0]) 
    .to_dict()
)
cluster_color_map = {
    cluster: party_color_map[party] for cluster, party in cluster_party_map.items()
}

plt.figure(figsize=(20, 6))

plt.subplot(1, 2, 1)
for parti in unique_parties:
    subset = df[df['parti'] == parti]
    plt.scatter(subset['PCA1'], subset['PCA2'], label=parti,
                color=party_color_map[parti], s=100, alpha=0.7)
plt.title("Actual parties")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
plt.grid(True)

# Plot B: KMeans Clusters with matched colors
plt.subplot(1, 2, 2)
for cluster_id in df['Cluster'].unique():
    subset = df[df['Cluster'] == cluster_id]
    plt.scatter(subset['PCA1'], subset['PCA2'], label=f"Cluster {cluster_id}",
                color=cluster_color_map[cluster_id], s=100, alpha=0.7)
plt.title("KMeans Clustering")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.grid(True)

plt.tight_layout()
plt.show()

ari = adjusted_rand_score(df['parti'], df['Cluster']) #Tells us how well the clustering matches the actual party labels.
nmi = normalized_mutual_info_score(df['parti'], df['Cluster']) #Tells us how well knowing one set of lables helps predict the other set.
print(f"Adjusted Rand Index (ARI): {ari:.3f}")
print(f"Normalized Mutual Information (NMI): {nmi:.3f}")

#Here we have made the clustering with 14 centroids to mach the parties.
```

## How K-Means Clustering works

You pick k, the number of clusters you want.

It randomly places k centroids (cluster centers).

Each data point is assigned to the nearest centroid.

Then centroids move to the mean position of their assigned points.

This repeats until assignments stabilize.

## Characteristics

You must specify k upfront.

Works well with roundish, equally sized clusters.

Sensitive to initial centroid placement.

Efficient and fast on large datasets.

```python
inertias = []
k_range = range(1, 15)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

plt.plot(k_range, inertias, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.grid(True)
plt.show()
```

Here we have plotted it into a graph to see the elbow point, which is the point where the inertia starts to decrease at a slower rate. This is the optimal number of clusters.

Inertia is the average squared distance from the centroid to each points in the plot.

## How Hierarchical Clustering works

Doesn’t require number of clusters in advance.

Builds a tree (dendrogram) showing data grouped step-by-step.

Two main types:

Agglomerative (bottom-up): start with each point as its own cluster, then merge the closest pairs stepwise.

Divisive (top-down): start with all points in one cluster, then split stepwise.

You "cut" the tree at the height you want to get your clusters.

## Characteristics

No need to specify k beforehand.

Good for discovering nested cluster structure.

Can be computationally expensive on big datasets.

Results are easy to interpret via dendrogram.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import dendrogram, ward

df = pd.read_excel("alldata.xlsx")

exclude_columns = ['navn', 'parti', 'storkreds', 'alder']
question_columns = [col for col in df.columns if col not in exclude_columns]
X = df[question_columns]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

linkage_array = ward(X_scaled)

plt.figure(figsize=(10, 6))
dendrogram(linkage_array)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.show()

#We make the dendrogram and can then see where the distances begin to be larger and use it to find the optimal number of clusters.
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_excel("alldata.xlsx")

exclude_columns = ['navn', 'parti', 'storkreds', 'alder']
question_columns = [col for col in df.columns if col not in exclude_columns]
X = df[question_columns]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

def init_nth_lowest(n):
    return np.full(n, np.inf).tolist()

def add_to_nth_lowest(lowest, value):
    lowest.append(value)
    lowest.sort()
    lowest.pop()

def nth_lowest_distance(n, point, points):
    lowest = init_nth_lowest(n)
    for p in points:
        distance = np.linalg.norm(point - p)
        add_to_nth_lowest(lowest, distance)
    return lowest[-1]

def nth_lowest_distances(n, points):
    return [nth_lowest_distance(n, point, points) for point in points]

# Calculate the 4th nearest neighbor distances (min_samples is often set between to 3 to 5 in DBSCAN when 2D but for this it would be 50 dimensions based on amount of questions)
distances = nth_lowest_distances(50, X_scaled)

distances.sort(reverse=True)

plt.figure(figsize=(10, 6))
plt.plot(range(len(distances)), distances)
plt.title("K-distance Plot for DBSCAN")
plt.xlabel("Points (Candidates)")
plt.ylabel("Distance to 50th nearest neighbor (Eps)")
plt.grid(True)
plt.show()

#Here we have plotted the distances to the 4th nearest neighbor, so that we can find the elbow point which gives a optimal value for eps in DBSCAN.
```

## How DBSCAN works

Groups points that are densely packed together.

Requires two parameters:

eps: max distance between points to be neighbors.

min_samples: minimum points required to form a dense region.

Points in dense regions form clusters.

Points that are far from any cluster are labeled noise or outliers.

### Characteristics

Does not require number of clusters upfront.

Can find clusters of any shape (not just round).

Robust to noise and outliers.

Can struggle with varying density clusters or high-dimensional data.




### Diemension Changes

The shapes it marks and the outliers looks off becuase we reduce in dimensionality after

## An overview of the political landscape of the elected candidates, highlighting which members agree or disagree the most and which parties or party members have significant disagreements.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity

df = pd.read_excel("alldata.xlsx")

exclude_columns = ['navn', 'parti', 'storkreds', 'alder']
question_columns = [col for col in df.columns if col not in exclude_columns]
X = df[question_columns].copy()
X = X.apply(pd.to_numeric, errors='coerce')
X = X.fillna(X.median()) 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)

loadings = pd.DataFrame(pca.components_.T, index=question_columns, columns=["PC1", "PC2"])
important_PC1 = loadings["PC1"].abs().sort_values(ascending=False)
important_PC2 = loadings["PC2"].abs().sort_values(ascending=False)

pca_df = pd.DataFrame(principal_components, columns=["PC1", "PC2"])
pca_df["parti"] = df["parti"]
pca_df["navn"] = df["navn"].str.strip()

#Cosine similarity between politicians
similarity_matrix = cosine_similarity(X_scaled)
similarity_df = pd.DataFrame(similarity_matrix, index=df['navn'].str.strip(), columns=df['navn'].str.strip())  # Strip spaces in names

#Removes semilarities with themselves.
similarity_df_no_diag = similarity_df.copy() 
for i in similarity_df_no_diag.index:
    similarity_df_no_diag.loc[i, i] = None

stacked = similarity_df_no_diag.stack() #for ease

# Remove duplicates by keeping only unique pairs (unordered) (A, B) and (B, A).
stacked_unique = stacked[stacked.index.get_level_values(0) < stacked.index.get_level_values(1)]

stacked_sorted = stacked_unique.sort_values()

print("\nMost Disagreement Between Politicians:")
print(stacked_sorted.head(10))

print("\nMost Agreement Between Politicians:")
print(stacked_sorted[::-1].head(10))

#Here is shown the most agreement and disagreement between the politicians based on the cosine similarity of their answers to the questions.
```

| Concept                    | Meaning                                                         |
| -------------------------- | --------------------------------------------------------------- |
| Each politician = a vector | A list of their answers (a point in multi-dimensional space)    |
| Cosine similarity          | Measures how **parallel** their answer patterns are             |
| Low similarity (near -1)   | Answers are **opposite in direction** → **strong disagreement** |
| High similarity (near 1)   | Answers are **parallel** → **strong agreement**                 |

Cosine similarity then captures how ideologically aligned two people are — without being overly sensitive to how "strongly" they answered (magnitude is normalized away).

each politician’s answers become a standardized vector (mean 0, std 1).
